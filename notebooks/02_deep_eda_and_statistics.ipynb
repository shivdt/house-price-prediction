### **Mutual Information**

* The mutual information (MI) between two quantities is a measure of the extent to which knowledge of one quantity reduces uncertainty about the other. If you knew the value of a feature, how much more confident would you be about the target?

* As MI is a univariate metric, it can't detect interactions between features.

* However, high MI does not guarantee that a feture will have greater impact on model predictions, it depends on type of model we chose and also requre domain knowledge to expose hidden association which can be easily learned by models. This is because MI is the dependence between feature and target. 

* Thus we can use MI to get a rough idea about important features to begin with, along with correlation (which detects linear relationships), rather than randomly picking up features for analysis.

* <u>**Note**</u>: scikit-learn has two separate main functions: mutual_info_regression (for a continuous target like SalePrice) and mutual_info_classif (for a discrete/categorical target)

Example: ExterQual vs SalePrice

```python
qual_order = ['Ex', 'Gd', 'TA', 'Fa']

plt.figure(figsize=(8, 5))
sns.boxplot(x='ExterQual', y='SalePrice', data=train_df, order=qual_order)

plt.title('Impact of Exterior Quality on Sale Price')
plt.xlabel('Exterior Quality')
plt.ylabel('Sale Price ($)')
plt.show()
```

* Knowing the exterior quality of a house reduces uncertainty about its sale price.

* factorize() actually returns two things: codes, uniques = X[colname].factorize()

Example:

    codes:
        [0, 1, 0, -1, 1]

    uniques:
        ["CollgCr", "Veenker"]

```python
from sklearn.feature_selection import mutual_info_regression


# Make local copies
X = train_df.copy().drop(["Id", "SalePrice"], axis=1)
y = train_df["SalePrice"].copy()

# Factorize text categories into dummy integers
for colname in X.select_dtypes(include=["object", "string"]): # included string as pandas object backward compatibility
    # pd.factorize handles missing values by assigning them an integer index of -1
    X[colname], _ = X[colname].factorize()



# Fill missing numerical values with column medians
X = X.fillna(X.median())



discrete_features = (X.dtypes == int)



# Calculate mutual information
mi_scores = mutual_info_regression(
    X,
    y,
    discrete_features=discrete_features,
    random_state=42
)

# Create Series
mi_scores = (
    pd.Series(mi_scores, name="MI Scores", index=X.columns)
    .sort_values(ascending=True)
)

# Plot
plt.figure(figsize=(10, len(mi_scores) * 0.35)) # For enough space

sns.barplot(
    x=mi_scores.values,
    y=mi_scores.index
)

plt.title("Mutual Information Score")
plt.xlabel("Mutual Information Score")
plt.ylabel("Features")
plt.show()
```

## Check Skewness

* I would like to use the convention:
    - |skew| < 0.5 → fairly symmetric
    - 0.5 ≤ |skew| < 1 → moderately skewed
    - |skew| ≥ 1 → heavily skewed → may need transformation

#### Get Skewness

```python
# Calculate skewness
skewness = (
    train_df[numerical_cols]
    .skew()
    .sort_values(key=abs, ascending=False)
)

#print(skewness)

highly_skewed = skewness[abs(skewness) >= 1]

print("Number of highly skewed features:", len(highly_skewed))
print(highly_skewed)
```

```python
highly_skewed_features = highly_skewed.index

null_counts = train_df[highly_skewed_features].isnull().sum()
print("Null value counts in highly skewed features:\n", null_counts)


print()

zero_counts = (train_df[highly_skewed_features] == 0).sum()
print("Zero counts in highly skewed features:\n", zero_counts)
```

> **Note:** As most of them have lots of zeros, using log(1+x) is preferred.
Most of the features appear skewed due lot of null values or zeros, so only few feature have legitimate high skewness like LotArea, TotalBsmtSF, 1stFlrSF, GrLivArea.  

> **Note:** We do not apply log transformation to categorical variables

#### Make a Skew Table

```python
skew_table = pd.DataFrame(
    {
        "Skewness": train_df[numerical_cols].skew(),
        "Abs_skewness": train_df[numerical_cols].skew().abs()
    }
).sort_values("Abs_skewness", ascending=False)

print(skew_table)
```

#### Conclusion

> **Note:** As tree-based models such as Random Forest, XGBoost, LightGBM, etc. are generally much less sensitive to feature skewness than linear models, and we are going to use tree based models only so we can safely skip transformation of skewed features (log transformation can sometimes make a nonlinear relationship more linear but as tree based models calculate threshold non linearity does not matter). 

> But the Kaggle competition mentions the following: "Metric: Submissions are evaluated on Root-Mean-Squared-Error (RMSE) between the logarithm of the predicted value and the logarithm of the observed sales price.". So even though tree based models like XGBoost not require log-transformation, I think we ust train our model on log transformed SalePrice based on evaluation metric.

## LandContour vs LandSlope

```python
# Ai Generated
import pandas as pd
import scipy.stats as ss

def cramers_v(series1, series2):
    # 1. Create contingency table
    contingency_matrix = pd.crosstab(series1, series2)
    
    # 2. Compute Chi-Square statistic
    chi2 = ss.chi2_contingency(contingency_matrix)[0]
    n = contingency_matrix.sum().sum()
    
    # 3. Calculate Cramer's V formula
    phi2 = chi2 / n
    r, k = contingency_matrix.shape
    return (phi2 / min(k - 1, r - 1)) ** 0.5

# Calculate correlation
correlation = cramers_v(train_df['LandContour'], train_df['LandSlope'])
print(f"Cramér's V Correlation between LandContour and LandSlope: {correlation:.4f}")
```

```python
# Frequency table (contingency table)
crosstab_counts = pd.crosstab(train_df['LandContour'], train_df['LandSlope'], margins=True)
print(crosstab_counts)
```

```python
# Proportion table (Percentages across rows)
crosstab_pct = pd.crosstab(
    train_df['LandContour'], 
    train_df['LandSlope'], 
    normalize='index'
) * 100
# The normalize='index' parameter divides each value in a row by the sum of that row,

print(crosstab_pct.round(2))
```
